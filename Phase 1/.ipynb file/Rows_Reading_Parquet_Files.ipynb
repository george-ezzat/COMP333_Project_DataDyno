```python
import pyarrow.parquet as pq
import os
```

```python
folder_path = r"C:\Users\HP HOME\.cache\kagglehub\datasets\arvindnagaonkar\flight-delay\versions\2"
```

```python
parquet_files = ["features_added.parquet", "Flight_Delay.parquet"]
```

```python
for file_name in parquet_files:
    parquet_file = os.path.join(folder_path, file_name)
    table = pq.ParquetFile(parquet_file)
    num_rows = table.metadata.num_rows
    num_columns = len(table.schema)  # Count of columns
    print(f"{file_name} -> Rows: {num_rows}, Columns: {num_columns}")
```
